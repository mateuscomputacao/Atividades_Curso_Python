# Aula 09: Funções

Até hoje, nós escrevemos o código de cima para baixo. Mas e se precisarmos repetir um mesmo bloco de código em várias partes do programa (por exemplo, limpar a tela e mostrar um menu)?
As **Funções** nos permitem dar um *nome* para um bloco de código, para que possamos "chamá-lo" quando quisermos, ajudando a organizar nosso programa.

## 1. Criando uma Função

Para "definir" uma função, usamos a palavra `def`, seguida do nome da função e parênteses `()`.

```python
def dar_boas_vindas():
    print("----------------------")
    print(" Bem vindo ao Sistema ")
    print("----------------------")

# A função não faz nada até ser explicitamente "chamada"
print("Iniciando...")
dar_boas_vindas()
print("Fazendo algo no sistema...")
dar_boas_vindas()
```

## 2. Passando Parâmetros (Argumentos)

Funções podem receber dados de fora para trabalhar com eles. Chamamos isso de "parâmetros".

```python
def saudar_pessoa(nome_da_pessoa):
    print(f"Olá, {nome_da_pessoa}! Seja muito bem-vindo.")

saudar_pessoa("Maria")
saudar_pessoa("Pedro")
```

Podemos ter múltiplos parâmetros, separados por vírgula.

```python
def somar(a, b):
    resultado = a + b
    print(f"A soma de {a} com {b} é {resultado}")

somar(10, 5)
somar(100, 200)
```

## 3. Retornando Valores (`return`)

A função `somar` acima imprimiu o valor na tela. Mas e se quiséssemos pegar esse resultado para usar em outra conta? Funções podem **devolver** um resultado para quem as chamou usando a palavra `return`.

```python
def multiplicar(a, b):
    resultado = a * b
    return resultado # "Devolve" o valor, mas não imprime!

# Guardamos o retorno em uma variável
dobro_de_10 = multiplicar(10, 2)
print(f"O retorno da função foi {dobro_de_10}")

# Assim podemos usar o resultado em outra parte do código
print(f"O quádruplo de 10 é {dobro_de_10 * 2}")
```

## 4. Retornando Múltiplos Valores (Tuplas)

Uma característica incrível e muito usada no Python é a capacidade de uma função retornar **mais de um valor ao mesmo tempo**. Diferente de linguagens como C ou Java (onde você normalmente tem que fazer truques ou criar classes para devolver dois valores), no Python basta separá-los por vírgula no `return`.

```python
def calcular_estatisticas(numeros):
    # Calcula três coisas de uma vez
    total = sum(numeros)
    maior = max(numeros)
    menor = min(numeros)

    # Retornamos os três valores separados por vírgula.
    # Isso magicamente os agrupa no que o Python chama de 'Tupla' (parecida com uma lista, mas imutável)
    return total, maior, menor

minhas_notas = [7, 8, 10, 5, 9]

# Na hora de chamar a função, também usamos variáveis separadas por vírgula para receber as 3 respostas!
# Chamamos isso de 'Desempacotamento' (Unpacking)
soma_total, nota_maxima, nota_minima = calcular_estatisticas(minhas_notas)

print(f"Soma: {soma_total}")
print(f"Máxima: {nota_maxima}")
print(f"Mínima: {nota_minima}")
```

Se você guardar a resposta numa variável só (sem desempacotar), o Python te devolverá os itens entre parênteses `()`. Isso é o que define uma **Tupla**.

```python
tudo_junto = calcular_estatisticas(minhas_notas)
print("Retorno direto (Tupla):", tudo_junto)

# Tuplas funcionam igual a listas para acessar pelo índice, mas você não pode alterá-las depois de criadas:
print("Acessando o primeiro item da tupla (soma):", tudo_junto[0])
```

### Desafio 6: Contar Vogais
Crie uma função chamada `contar_vogais` que recebe uma string como parâmetro e retorna o número de vogais (a, e, i, o, u, maiúsculas e minúsculas) presentes na string.

```python
def contar_vogais(palavra):
  vogais = "aeiouAEIOU"
  qtd_vogais = 0
  for char in palavra:
    if char in vogais:
      qtd_vogais += 1
  return qtd_vogais

frase = input("Digite uma palavra ou frase: ")
print(f"A frase '{frase}' tem {contar_vogais(frase)} vogais.")
```

### Desafio 7: Inverter String
Crie uma função chamada `inverter_string` que recebe uma string e retorna a string invertida. Não use fatiamento (`[::-1]`).

```python
# Seu código aqui

def inverter_string(palavra):
  string_invertida = ""
  for i in range(len(palavra) - 1, -1, -1):
    string_invertida += palavra[i]
  return string_invertida

palavra_original = input("Digite uma palavra: ")
print(f"A palavra original é: {palavra_original}.\nA palavra invertida é: {inverter_string(palavra_original)}")
```

### Desafio 8: Verificar Palíndromo
Crie uma função chamada `eh_palindromo` que recebe uma string e retorna `True` se ela for um palíndromo (lê-se da mesma forma de trás para frente), e `False` caso contrário. Ignore espaços e diferenças de maiúsculas/minúsculas.

```python
# Seu código aqui

def eh_palindromo(palavra):
  palavra_invertida = ""
  for i in range(len(palavra) - 1, -1, -1):
    palavra_invertida += palavra[i]
  if palavra_invertida == palavra:
    return True
  else:
    return False

palavra = input("Digite uma palavra: ").lower().replace(" ", "")
if eh_palindromo(palavra):
  print(f"A palavra {palavra} é um palíndromo.")
else:
  print(f"A palavra {palavra} não é um palíndromo.")

```

### Desafio 9: Fatorial
Crie uma função chamada `calcular_fatorial` que recebe um número inteiro não negativo e retorna o seu fatorial. O fatorial de um número `n` (representado por `n!`) é o produto de todos os inteiros positivos menores ou iguais a `n`. Por exemplo, `5! = 5 * 4 * 3 * 2 * 1 = 120`. Utilize um laço de repetição.

```python
# Seu código aqui

def calcular_fatorial(n):
  fatorial = 1
  for i in range(n, 1, -1):
    fatorial = fatorial*i
  return fatorial

while True:
    try:
        numero = int(input("Digite um número positivo maior que zero: "))

        if numero < 1:
            print("Número inválido!")
        else:
            break
    except ValueError:
        print("Entrada inválida.")

print(f"\nO fatorial de {numero} é {calcular_fatorial(numero)}.")
```

### Desafio 10: Sequência de Fibonacci
Crie uma função chamada `fibonacci_sequence` que recebe um número `n` e retorna uma lista com os primeiros `n` números da sequência de Fibonacci. A sequência começa com 0 e 1, e cada número subsequente é a soma dos dois anteriores (ex: 0, 1, 1, 2, 3, 5, 8...).

```python
def sequencia_fibonacci(n):
  if n <= 0:
    return []
  elif n == 1:
    return [0]
  else:
    sequencia = [0, 1]
    while len(sequencia) < n:
      proximo_numero = sequencia[-1] + sequencia[-2]
      sequencia.append(proximo_numero)
    return sequencia

termo = int(input("Digite um núemero n para ver a \nSequência de Fibonacci até o termo N: "))
print(f"\nOs primeiros {termo} números de Fibonacci são: {sequencia_fibonacci(termo)}")
```

### Desafio 11: Remover Duplicatas
Crie uma função chamada `remover_duplicatas` que recebe uma lista e retorna uma nova lista com todos os elementos duplicados removidos, mantendo a ordem original dos elementos que restaram.

```python
# Seu código aqui

def remover_duplicatas(lista):
  lista_unicos = []
  qtd_numeros_lista = len(lista)
  for i in range(qtd_numeros_lista):
    flag = True
    qtd_numeros_unicos = len(lista_unicos)
    for j in range(qtd_numeros_unicos):
      if lista[i] == lista_unicos[j]:
        flag = False
        break
    if flag == True:
      lista_unicos.append(lista[i])
  return lista_unicos

lista_duplicatas = [5, 5, 5, 5, 7, 7, 8, 9, 5, 2, 7, 8, 4, 5, 1, 2, 3]
remover_duplicatas(lista_duplicatas)

```

### Desafio 12: Verificar Existência de Elemento
Crie uma função chamada `contem_elemento` que recebe uma lista e um `elemento` e retorna `True` se o `elemento` estiver na lista, e `False` caso contrário. Não use o operador `in` em sua implementação.

```python
# Seu código aqui

def contem_elemento(lista, elemento):
  cont = 0
  flag = False
  while cont < len(lista):
    if lista[cont] == elemento:
      flag = True
      break
    else:
      cont += 1
  return flag

lista = ["Banana", "Melão", "Maçã", "Abacate", "Pêra"]
elemento = "Abacate"

contem_elemento(lista, elemento)
```
